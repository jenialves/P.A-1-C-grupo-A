```python
# exercício 9

# crie uma lista vazia chamada números
# peça ao usuário para informar 10 números
# depois crie duas listas vazias: pares e ímpares
# percorra a lista original e coloque cada número na lista correspondente

numeros = [ ]
pares = [ ]
ímpares = [ ]

for i in range(10):
  numeros.append(int(input("Informe um número: ")))

  if numeros [i] % 2 == 0:
    pares.append(numeros[i])

  else:
    ímpares.append(numeros[i])

print('lista inteira', numeros)
print('números pares', pares)
print('números ínpares', ímpares)
```

```python
# exercício 10

# faça um pequeno sistema para cadastrar as notas de 5 alunos
# o programa deverá armazenar as notas em uma lista e depois mostrar :

# todas as notas
# maior nota
# média da turma
# quantidade de alunos com nota maior ou igual a 6
# quantidade de alunos com nota menos que 6

notas = []
for i in range(1,6):
  nota = float(input('Digite a nota do aluno: '))
  notas.append(nota)

maior6 = len([i for i in notas if i >= 6])
menor6 = len([i for i in notas if i < 6])

print(f'Todas as notas: {notas}')
print(f'Maior nota: {max(notas)}')
mediaturma = sum(notas)/len(notas)
print(f'Média da turma: {mediaturma:.2f}')
print(f'Quantidade de alunos com média maior ou igual a 6: {maior6}')
print(f'Quantidade de alunos com média menor que 6: {menor6}')
```
